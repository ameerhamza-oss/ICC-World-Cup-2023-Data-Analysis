```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries Imported Successfully!")
```

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x="team",
    order=df["team"].value_counts().index
)

plt.title("Number of Records by Team")
plt.xticks(rotation=45)

plt.show()
```

```python
top_runs = df.groupby("player")["runs"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,5))
top_runs.plot(kind="bar")

plt.title("Top 10 Run Scorers - ICC World Cup 2023")
plt.xlabel("Player")
plt.ylabel("Total Runs")

plt.xticks(rotation=45)
plt.show()
```

```python
top_runs
```

```python
top_wickets = df.groupby("player")["wkts"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(12,5))
top_wickets.plot(kind="bar")

plt.title("Top 10 Wicket Takers - ICC World Cup 2023")
plt.xlabel("Player")
plt.ylabel("Total Wickets")

plt.xticks(rotation=45)
plt.show()
```

```python
team_runs = df.groupby("team")["runs"].sum().sort_values(ascending=False)

plt.figure(figsize=(12,5))

team_runs.plot(kind="bar")

plt.title("Total Runs by Team")
plt.xlabel("Team")
plt.ylabel("Total Runs")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(12,8))

sns.heatmap(
    df.select_dtypes(include=["int64", "float64"]).corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
df["start_date"] = pd.to_datetime(df["start_date"])

print(df["start_date"].dtype)
```

```python
batting_df = df[df["bat_or_bowl"] == "bat"]

bowling_df = df[df["bat_or_bowl"] == "bowl"]

print("Batting Records:", batting_df.shape)
print("Bowling Records:", bowling_df.shape)
```

```python
top_avg = batting_df.groupby("player")["runs"].mean().sort_values(ascending=False).head(10)

top_avg
```

```python
top_econ = bowling_df.groupby("player")["econ"].mean().sort_values().head(10)

top_econ
```
